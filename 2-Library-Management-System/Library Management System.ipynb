# Project 2: LIBRARY MANAGEMENT SYSTEM

## 1. Book

```python
class Book:
    def __init__(self, book_Id, title, author, category):
        self.book_Id = book_Id
        self.title = title
        self.author = author
        self.category = category
        self.status = "Available"

    
    def display_book(self):
        print("Book ID:",self.book_Id)
        print("Title:", self.title)
        print("Author:", self.author)
        print("Category:",self.category)
        print("Status:",self.status)

    
    def issue_book(self):
        if self.status == "Available":
            self.status = "Issued"
            return True
        else:
            return False

    
    def return_book(self):
        self.status = "Available"
```

## 2. Member

```python
class Member:
    def __init__(self, member_Id, name):
        self.member_Id = member_Id
        self.name = name
        self.borrowed_books = []

    
    def display_member(self):
        print("Member Id:",self.member_Id)
        print("Name:", self.name)
        print("Borrowed books:", self.borrowed_books)

    
    def borrow_book(self, book_Id):
        if book_Id not in self.borrowed_books:
            self.borrowed_books.append(book_Id)

    
    def return_book(self, book_Id):
        if book_Id in self.borrowed_books:
            self.borrowed_books.remove(book_Id)
```

## 3. Library

```python
class Library:
    def __init__(self):
        self.books = []
        self.members = []

    # -----BOOK FUNCTION-----
    
    def add_book(self, book):
        for existing_book in self.books:
            if existing_book.book_Id == book.book_Id:
                print("Book ID already exists!")
                return
                
        self.books.append(book)
        print("Book added successfully!")

    
    def remove_book(self, book_Id): 
        for book in self.books:
            if book.book_Id == book_Id: 
                self.books.remove(book)
                print("Book removed successfully!")
                return
        print("Book not found")

    
    def search_book(self, book_Id): 
        for book in self.books: 
            if book.book_Id == book_Id: 
                return book
        return None

    
    def display_books(self): 
        if len(self.books) == 0: 
            print("No books available.")
        else:
            for book in self.books: 
                print("------------------")
                book.display_book()

    # -----MEMBER FUNCTION-----
    
    def add_member(self, member):
        for existing_member in self.members:
            if existing_member.member_Id == member.member_Id: 
                print("Member ID already exists!")
                return
                
        self.members.append(member)
        print("Member added successfully!")

    
    def remove_member(self, member_Id): 
        for member in self.members: 
            if member.member_Id == member_Id: 
                self.members.remove(member)
                print("Member removed successfully!")
                return 
        print("Member not found!")

    def search_member(self, member_Id): 
        for member in self.members:
            if member.member_Id == member_Id: 
                return member
        return None

    def display_members(self): 
        if len(self.members) == 0: 
            print("No member registered.")
        else: 
            for member in self.members: 
                print("------------------")
            member.display_member()

    # -----ISSUE BOOK-----
    
    def issue_book(self, book_Id, member_Id): 
        book = self.search_book(book_Id)
        member = self.search_member(member_Id)

        if book is None: 
            print("Book not found!")
            return

        if member is None: 
            print("Member not found!")
            return 
            
        if book.issue_book(): 
            member.borrow_book(book_Id)
            print("Book issued successfully!")
        else: 
            print("Book is already issued!")

    # -----RETURN BOOK----- 
    
    def return_book(self, book_Id, member_Id): 
        book = self.search_book(book_Id)
        member = self.search_member(member_Id)

        if book is None: 
            print("Book not found!")
            return
            
        if member is None: 
            print("Member not found!")
            return

        if book_Id not in member.borrowed_books:
            print("This member has not borrowed this book.")
            return
            
        book.return_book()
        member.return_book(book_Id)

        print("Book returned successfully.")
```

## 4. Library Management System

```python
class LibraryManagementSystem: 
                
    def __init__(self):
        self.library = Library()
                
    def display_menu(self):
        print("\n------LIBRARY MANAGEMENT SYSTEM-----")
        print("1. Add Book")
        print("2. Remove Book")
        print("3. Search Book")
        print("4. Display Books")
        print("5. Add Member")
        print("6. Remove Member")
        print("7. Search Member")
        print("8. Display Member")
        print("9. Issue Book")
        print("10. Return Book")
        print("11. Exit")
                
    def run(self): 
                
        while True:
            self.display_menu()
            
            choice = input("Enter your choice: ")
                
            # -----ADD BOOK-----
            if choice == "1":
                try:
                    book_id = int(input("Enter Book ID: "))
                except ValueError:
                    print("Book ID must be a number.")
                    continue
                    
                title = input("Enter Title: ")
                author = input("Enter Author: ")
                category = input("Enter Category: ")
                
                book = Book(book_id, title, author, category)
                self.library.add_book(book)
            
            # -----REMOVE BOOK-----
            elif choice == "2":
                
                book_id = int(input("Enter Book ID: "))
                
                self.library.remove_book(book_id)
                
            # -----SEARCH BOOK-----
            elif choice == "3":  
                
                book_id = int(input("Enter Book ID: "))
                book = self.library.search_book(book_id)
                
                if book:
                    book.display_book()
                else:
                    print("Book not found!")
                
            # -----DISPLAY BOOKS-----
            elif choice == "4":
                
                self.library.display_books()
                
            # -----ADD MEMBER----- 
            elif choice == "5":
                
                member_id = int(input("Enter Member ID: "))
                name = input("Enter Member Name: ")
                
                member = Member(member_id, name)
                
                self.library.add_member(member)
                
            # -----REMOVE MEMBER-----
            elif choice == "6":
                
                member_id = int(input("Enter Member ID: "))
                
                self.library.remove_member(member_id)
                
            # -----SEARCH MEMBER-----
            elif choice == "7":
                
                member_id = int(input("Enter Member ID: "))
                member = self.library.search_member(member_id)
                
                if member:
                    member.display_member()
                else:
                    print("Member not found!")
                
            # -----DISPLAY MEMBERS-----
            elif choice == "8":
                
                self.library.display_members()
                
            # -----ISSUE BOOK-----
            elif choice == "9":
                
                book_id = int(input("Enter Book ID: "))
                member_id = int(input("Enter Member ID: "))
                
                self.library.issue_book(book_id, member_id)
                
            # -----RETURN BOOK-----
            elif choice == "10":
                
                book_id = int(input("Enter Book ID: "))
                member_id = int(input("Enter Member ID: "))
                
                self.library.return_book(book_id, member_id)
                
            # -----EXIT-----
            elif choice == "11":
                
                print("Thank you for using the Library Management System!")
                break
                
            else:
                print("Invalid choice!")

# START PROGRAM
system = LibraryManagementSystem()
system.run()
```
